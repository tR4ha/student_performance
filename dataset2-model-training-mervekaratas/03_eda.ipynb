```python
import pandas as pd

import matplotlib.pyplot as plt

import seaborn as sns
```

```python
df = pd.read_csv(
    "../data/raw/Student_performance_data _ (1).csv"
)
```

```python
plt.figure(figsize=(6,4))

sns.countplot(
    x=df["GradeClass"]
)

plt.title(
    "Distribution of GradeClass"
)

plt.show()
```

```python
plt.figure(figsize=(6,4))

sns.histplot(
    df["GPA"],
    kde=True
)

plt.title(
    "Distribution of GPA"
)

plt.show()
```

```python
plt.figure(figsize=(14,10))

sns.heatmap(
    df.corr(numeric_only=True),
    annot=True,
    cmap="coolwarm"
)

plt.title(
    "Correlation Matrix"
)

plt.show()
```

```python
categorical_cols = [
    "Gender",
    "Ethnicity",
    "ParentalEducation",
    "Tutoring",
    "ParentalSupport",
    "Extracurricular",
    "Sports",
    "Music",
    "Volunteering"
]

fig, axes = plt.subplots(
    nrows=3,
    ncols=3,
    figsize=(18,12)
)

axes = axes.flatten()

for i, col in enumerate(categorical_cols):

    sns.boxplot(
        x=df[col],
        y=df["GPA"],
        ax=axes[i]
    )

    axes[i].set_title(
        f"GPA by {col}"
    )

plt.tight_layout()

plt.show()
```

```python
plt.figure(figsize=(6,4))

sns.scatterplot(
    x=df["StudyTimeWeekly"],
    y=df["GPA"]
)

plt.title(
    "Study Time vs GPA"
)

plt.show()
```

```python
plt.figure(figsize=(6,4))

sns.scatterplot(
    x=df["Absences"],
    y=df["GPA"]
)

plt.title(
    "Absences vs GPA"
)

plt.show()
```

```python
plt.figure(figsize=(6,4))

sns.boxplot(
    x=df["GradeClass"],
    y=df["GPA"]
)

plt.title(
    "GradeClass vs GPA"
)

plt.show()
```

```python
sns.pairplot(
    df[[
        "GPA",
        "StudyTimeWeekly",
        "Absences",
        "GradeClass"
    ]]
)

plt.show()
```

```python
import scipy.stats as stats

plt.figure(figsize=(6,4))

stats.probplot(
    df["GPA"],
    dist="norm",
    plot=plt
)

plt.title("Probability Plot of GPA")

plt.show()
```
